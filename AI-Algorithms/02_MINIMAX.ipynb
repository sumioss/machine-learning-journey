# *Minimax Pseudocode*

FUNCTION MINIMAX(current_node, depth, is_maximizing):

    IF depth = 0 OR current_node has key 'value' THEN
        RETURN current_node.value OR 0
    END IF

    children ← current_node.children

    IF is_maximizing = TRUE THEN
        best ← -∞

        FOR each child IN children DO
            val ← MINIMAX(child, depth - 1, FALSE)
            best ← MAX(best, val)
        END FOR

        RETURN best

    ELSE
        best ← +∞

        FOR each child IN children DO
            val ← MINIMAX(child, depth - 1, TRUE)
            best ← MIN(best, val)
        END FOR

        RETURN best
    END IF

END FUNCTION

```python
def MINIMAX(current_node, depth, is_maximizing):

    if depth == 0 or 'value' in current_node:
        return current_node.get('value', 0)

    children = current_node['children']

    if is_maximizing:
        best = float('-inf')

        for child in children:
            val = MINIMAX(child, depth - 1, False)
            best = max(best, val)

        return best

    else:
        best = float('inf')

        for child in children:
            val = MINIMAX(child, depth - 1,True)
            best = min(best, val)

        return best
```

```python
tree = {
    'children': [
        {
            'children': [
                {'value': 3},
                {'value': 5}
            ]
        },
        {
            'children': [
                {'value': 6},
                {'value': 2}
            ]
        }
    ]
}

result = MINIMAX(tree, 2, True)
print("Final Minimax result:", result)
```
