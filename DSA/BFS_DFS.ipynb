```python
graph={
    'A':['B','C','D'],
    'B':['A','E','F'],
    'C':['A','F'],
    'D':['A'],
    'E':['B'],
    'F':['B','C']
}
start_node='A'
```

```python
def bfs(graph,start_node):
  v=[]
  q=[]
  q.append(start_node)
  while (q):
    current_node=q.pop(0)
    if current_node not in v:
      v.append(current_node)
      print(current_node)
      for node in graph[current_node]:
         if node not in v:
            q.append(node)
```

```python
bfs(graph,start_node)
```

```python
graph={
    'A':['B','C','D'],
    'B':['A','D'],
    'C':['A','D','E'],
    'D':['A','B','C','E'],
    'E':['C','D']
}
start_node='A'
```

```python
def dfs(graph,start_node):
  v=[]
  s=[]
  s.append(start_node)
  while len(s)!=0:
    current_node=s.pop(len(s)-1)
    if current_node not in v:
      v.append(current_node)
      print(current_node)
      for node in graph[current_node]:
        if node not in v:
          s.append(node)
```

```python
dfs(graph,start_node)
```
